## Variables en JavaScript

#### var (obsoleto)

```python
var nombre = "valor";
```

#### let (reasignable)

```python
let nombreVariable = 10;
```

#### const (no reasignable)

```python
const nombreConstante = "pi";
```

#### Mostrar variable en consola

```python
console.log(nombreVariable);
```

## Tipos de Datos

### Primitivos

#### String (Cadena de texto)

```python
const texto = "Hola mundo";
```

#### Number (Número)

```python
const entero = 42; 
const decimal = 3.14;
const infinito = Infinity;
const noNumero = NaN; // Not a Number
```

#### Boolean (Booleano)

```python
const verdadero = true;
const falso = false;
```

#### Null (valor nulo)

```python
const nulo = null;
console.log(nulo); // null
console.log(typeof nulo); // "object" (esto es un comportamiento histórico en JavaScript)
```

#### Undefined (Sin definir)

```python
let noDefinido;
const noDefinido2 = undefined;
```

## Datos No Primitivos (Objetos)

#### Array (Arreglo)

```python
const arreglo = [1, "dos", true, { clave: "valor" }];
console.log(arreglo[0]); // 1
```

#### Object (Objeto)

```python
const objeto = {
    clave: "valor",
    numero: 123,
    funcion: function () { } // Método del Objeto
}
```

#### Date (Fecha)

```python
const fecha = new Date()
```

### Comprobar Tipo de Dato:

```python
typeof "texto"; // "string"
typeof 42;      // "number"
typeof true;    // "boolean"
typeof null;    // "object"
typeof undefined; // "undefined"
typeof {};      // "object"
typeof [];      // "object"
typeof function () { }; // "function"
```
